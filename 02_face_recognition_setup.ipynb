# 02 - Face Recognition Setup & OpenCV Basics

Module A1 (OpenCV basics) + A3 (Face Recognition) demo, run against real registered photos.

```
import sys
sys.path.insert(0, '../app/services')
import cv2
import matplotlib.pyplot as plt
from cv_service import (to_grayscale, resize_image, apply_blur, detect_edges,
                         detect_faces, draw_face_boxes)

img = cv2.imread('../data/sample_faces/abhinav.jpg')
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
print("Original shape:", img.shape)
```

```
gray = to_grayscale(img)
blurred = apply_blur(gray, ksize=7)
edges = detect_edges(gray)
resized = resize_image(img_rgb, 224, 224)

fig, axes = plt.subplots(1, 4, figsize=(16,4))
axes[0].imshow(img_rgb); axes[0].set_title("Original")
axes[1].imshow(gray, cmap='gray'); axes[1].set_title("Grayscale")
axes[2].imshow(blurred, cmap='gray'); axes[2].set_title("Gaussian Blur")
axes[3].imshow(edges, cmap='gray'); axes[3].set_title("Canny Edges")
for ax in axes:
    ax.axis('off')
plt.tight_layout()
plt.savefig('cv_basics_demo.png', dpi=100)
plt.show()
```

### Face detection (Haar Cascade)

```
faces = detect_faces(img)
print(f"Faces detected: {len(faces)}")

boxed = draw_face_boxes(img_rgb, faces)
plt.figure(figsize=(5,5))
plt.imshow(boxed)
plt.title(f"{len(faces)} face(s) detected")
plt.axis('off')
plt.savefig('face_detection_demo.png', dpi=100)
plt.show()
```
